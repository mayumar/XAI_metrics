```python
import numpy as np


df_correlacion = X_train_phmd.drop(
    columns=["fault", "asset"]
)

corr = df_correlacion.corr(numeric_only=True)

# Conservamos solo el triángulo superior para evitar:
# - correlaciones de una variable consigo misma
# - pares duplicados, como (A, B) y (B, A)
mascara = np.triu(
    np.ones(corr.shape, dtype=bool),
    k=1,
)

correlaciones = (
    corr.where(mascara)
    .stack()
    .rename("correlacion")
    .reset_index()
    .rename(columns={
        "level_0": "variable_1",
        "level_1": "variable_2",
    })
)

correlaciones["valor_absoluto"] = correlaciones["correlacion"].abs()
```

```python
print(
    correlaciones[
        correlaciones["correlacion"].abs() >= 0.8
    ]
    .sort_values("valor_absoluto", ascending=False)
    .to_string(index=False)
)
```

```python
from config import DATASETS
metro_config = DATASETS["metro"]
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='raise')
df = df.set_index("timestamp").sort_index()
assert df.index.is_unique
df
```

```python
df = df.asfreq(metro_config['resample_freq'], method=metro_config['fill_method'])
df.index.name = "timestamp"
df
```

```python
df['failure'] = 0

for inicio, fin in metro_config['intervalos']:
    mascara = df.index.to_series().between(inicio, fin)
    df.loc[mascara, "failure"] = 1

df
```

```python
train_size = 0.56
val_size = 0.14
test_size = 0.30

n_total = len(df)

fin_train = int(n_total * train_size)
fin_val = int(n_total * (train_size + val_size))

train_dataset = df.iloc[:fin_train].copy()
val_dataset = df.iloc[fin_train:fin_val].copy()
test_dataset = df.iloc[fin_val:].copy()

X_train = train_dataset
y_train = train_dataset["failure"].rename("target")

X_val = val_dataset
y_val = val_dataset["failure"].rename("target")

X_test = test_dataset
y_test = test_dataset["failure"].rename("target")
```
