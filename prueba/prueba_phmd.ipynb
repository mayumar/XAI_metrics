## Calcular fracción de anomalías

```python
y_total = np.concatenate([
    np.asarray(y_train_phmd).ravel(),
    np.asarray(y_val_phmd).ravel(),
    np.asarray(y_test_phmd).ravel(),
])

anomalias_fraccion = np.mean(y_total == 1)

print("Fracción de anomalías:", anomalias_fraccion)
print("Porcentaje:", anomalias_fraccion * 100)
```

## Guardar dataset

```python
from pathlib import Path
import numpy as np
import pandas as pd

output_dir = Path("data") / "AHU21"
output_dir.mkdir(parents=True, exist_ok=True)

# Asegurar formatos consistentes
X_train_df = pd.DataFrame(X_train).drop(columns=['unit', 'fault'])
X_val_df = pd.DataFrame(X_val).drop(columns=['unit', 'fault'])
X_test_df = pd.DataFrame(X_test).drop(columns=['unit', 'fault'])

y_train_s = pd.Series(np.asarray(y_train_phmd).ravel(), name="target")
y_val_s = pd.Series(np.asarray(y_val_phmd).ravel(), name="target")
y_test_s = pd.Series(np.asarray(y_test_phmd).ravel(), name="target")

# Guardar sin índices para evitar columnas artificiales
X_train_df.to_csv(output_dir / "X_train.csv", index=False)
X_val_df.to_csv(output_dir / "X_val.csv", index=False)
X_test_df.to_csv(output_dir / "X_test.csv", index=False)

y_train_s.to_csv(output_dir / "y_train.csv", index=False)
y_val_s.to_csv(output_dir / "y_val.csv", index=False)
y_test_s.to_csv(output_dir / "y_test.csv", index=False)
```
